# 01. Exploratory Data Analysis: Mayo Clinic PBC Cohort
Comprehensive investigation of biomarker distributions, missingness patterns, and observational correlations with clinical outcomes.

```python
import sys
sys.path.insert(0, "..")
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_loader import load_raw_data
from src.validation import normalize_column_names, preprocess_target
from config.config import NUMERIC_FEATURES, CATEGORICAL_FEATURES, TARGET_CLASSES

df_raw = load_raw_data(auto_download=True)
df = normalize_column_names(df_raw)
y, dist_df = preprocess_target(df)
print(f"Cohort dimensions: {df.shape}")
dist_df
```

## 1. Missing Data Profiling

```python
missing_summary = pd.DataFrame({
    "Missing Count": df[NUMERIC_FEATURES + CATEGORICAL_FEATURES].isna().sum(),
    "Missing Pct (%)": (df[NUMERIC_FEATURES + CATEGORICAL_FEATURES].isna().sum() / len(df)) * 100
}).sort_values(by="Missing Pct (%)", ascending=False)
missing_summary[missing_summary["Missing Count"] > 0]
```

## 2. Physiological Biomarkers Stratified by Clinical Outcome

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
df_plot = df.copy()
df_plot["Outcome"] = y.map(TARGET_CLASSES)

sns.boxplot(data=df_plot, x="Outcome", y="Bilirubin", ax=axes[0, 0], palette="Set2")
axes[0, 0].set_title("Serum Bilirubin by Outcome")

sns.boxplot(data=df_plot, x="Outcome", y="Albumin", ax=axes[0, 1], palette="Set2")
axes[0, 1].set_title("Serum Albumin by Outcome")

sns.boxplot(data=df_plot, x="Outcome", y="Prothrombin", ax=axes[1, 0], palette="Set2")
axes[1, 0].set_title("Prothrombin Time by Outcome")

sns.boxplot(data=df_plot, x="Outcome", y="Copper", ax=axes[1, 1], palette="Set2")
axes[1, 1].set_title("Urine Copper by Outcome")

plt.tight_layout()
plt.show()
```
